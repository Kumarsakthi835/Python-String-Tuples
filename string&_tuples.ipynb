```python
# Task 1: String Concatenation
# String 1
string1 = "Hello"
# Get name from user
name = input("Enter your Name: ")
# Concatenate string1 and name
result1 = string1 + " " + name
print(result1)

# String 3
string3 = "Welcome to Python programming"

# Concatenate
result2 = string1 + " " + name + ", " + string3
print(result2)
```

```python
# Task 2: String Slicing and Indexing
# Using the concatenated string
string = "Hello Kumar, Welcome to Python programming"
# a. First character
print("a. First character:", string[0])
# b. Last character
print("b. Last character:", string[-1])
# c. First 5 characters
print("c. First 5 characters:", string[:5])
# d. Last 11 characters
print("d. Last 11 characters:", string[-11:])
# e. String in reverse
print("e. Reversed string:", string[::-1])
# f. Extract the word "Python"
print("f. Extract 'Python':", string[24:30])
```

```python
# Task 3: String Methods
# Input string
strM = "Python beginner tutorial"
# a. Convert to uppercase
print("a. Uppercase:", strM.upper())
# b. Convert to lowercase
print("b. Lowercase:", strM.lower())
# c. Capitalize
print("c. Capitalize:", strM.capitalize())
# d. Count occurrences of 't'
print("d. Count of 't':", strM.count('t'))
# e. Replace "Python" with "Data Analytics"
print("e. Replace:", strM.replace("Python", "Data Analytics"))
```

```python
# Task 4: Tuple Operations
# Create two tuples
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# a. Concatenate two tuples
t_combine = t1 + t2
print("a. Concatenated tuple:", t_combine)

# b. Repeat elements 3 times
print("b. Repeated 3 times:", t_combine * 3)

# c. Access 3rd element
print("c. 3rd element:", t_combine[2])

# d. First three elements
print("d. First three elements:", t_combine[:3])

# e. Last three elements
print("e. Last three elements:", t_combine[-3:])
```
